```python
import pandas as pd
from google.colab import files

uploaded = files.upload()
nama_file = list(uploaded.keys())[0]
df = pd.read_csv(nama_file)
print(df.head())

print("Ukuran dataset:", df.shape)
print("\nInformasi dataset:")
print(df.info())

print("Jumlah data kosong:")
print(df.isnull().sum())
print("\nJumlah data duplikat:")
print(df.duplicated().sum())

print(df["Grade"].value_counts())
print(df.drop_duplicates().shape)

df_bersih = df.drop_duplicates()
print("Distribusi Grade setelah menghapus duplikat:")
print(df_bersih["Grade"].value_counts())
print("\nPersentase Grade:")
print(df_bersih["Grade"].value_counts(normalize=True) * 100)

print(df.value_counts().head(10))

fitur = df.columns.drop("Grade")
cek_grade = df.groupby(list(fitur))["Grade"].nunique()
print("Jumlah kombinasi fitur dengan Grade berbeda:")
print((cek_grade > 1).sum())

print(df.describe())

print("Data dengan pH di bawah 6:")
print(df[df["pH"] < 6])
print("\nData dengan suhu di atas 60:")
print(df[df["Temprature"] > 60])

df_unik = df.drop_duplicates()

print("Sebelum menghapus duplikat:")
print("pH di bawah 6:", (df["pH"] < 6).sum())
print("Suhu di atas 60:", (df["Temprature"] > 60).sum())
print("\nSetelah menghapus duplikat:")
print("pH di bawah 6:", (df_unik["pH"] < 6).sum())
print("Suhu di atas 60:", (df_unik["Temprature"] > 60).sum())

print("Statistik pH berdasarkan Grade:")
print(df.groupby("Grade")["pH"].describe())

print("\nStatistik Temperature berdasarkan Grade:")
print(df.groupby("Grade")["Temprature"].describe())

print("Data pH di bawah 6:")
display(df[df["pH"] < 6])

print("Data suhu di atas 60:")
display(df[df["Temprature"] > 60])

print("pH di bawah 6:")
print(df[df["pH"] < 6]["Grade"].value_counts())

print("\nSuhu di atas 60:")
print(df[df["Temprature"] > 60]["Grade"].value_counts())

print(df.dtypes)

for kolom in df.columns:
    print(kolom, ":", df[kolom].unique())
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["Grade_encoded"] = le.fit_transform(df["Grade"])

print(df[["Grade", "Grade_encoded"]].drop_duplicates())

print("\nPemetaan Grade:")
for kategori, angka in zip(le.classes_, le.transform(le.classes_)):
    print(kategori, "=", angka)

X = df.drop(columns=["Grade", "Grade_encoded"])
y = df["Grade_encoded"]

groups = X.astype(str).agg("|".join, axis=1)

print("Jumlah fitur:", X.shape)
print("Jumlah target:", y.shape)

from sklearn.model_selection import StratifiedGroupKFold

sgkf = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

train_idx, test_idx = next(sgkf.split(X, y, groups))

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

print("Distribusi Grade Training:")
print(y_train.value_counts())

print("\nDistribusi Grade Testing:")
print(y_test.value_counts())

print("SELURUH DATA")
print(y.value_counts(normalize=True) * 100)

print("\nTRAINING")
print(y_train.value_counts(normalize=True) * 100)

print("\nTESTING")
print(y_test.value_counts(normalize=True) * 100)

print("=== PEMERIKSAAN TERBARU ===")

print("Jumlah kelompok unik:", groups.nunique())
print("Jumlah data training:", len(X_train))
print("Jumlah data testing:", len(X_test))

print("\nDistribusi training:")
print(y_train.value_counts().sort_index())

print("\nDistribusi testing:")
print(y_test.value_counts().sort_index())

print("\nData leakage:",
      set(groups.iloc[train_idx]).isdisjoint(
          set(groups.iloc[test_idx])
      ))
```
